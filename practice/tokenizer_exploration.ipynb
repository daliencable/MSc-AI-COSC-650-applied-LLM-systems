```python
"""
Supplementary exploration: why does GPT-4o compress Portuguese better than GPT-4?
Tests two hypotheses against words drawn from the Week 1 passage (Pessoa,
"Gosto de dizer", old Portuguese orthography):
  1. Diacritic-handling: o200k_base's regex keeps diacritics with their base
     letter; cl100k_base's doesn't.
  2. Vocabulary size: o200k_base (~200k tokens) covers more subwords than
     cl100k_base (~100k tokens), independent of diacritics.
Run with: python practice/tokenizer_exploration.py
"""

import tiktoken

gpt4 = tiktoken.get_encoding('cl100k_base')
gpt4o = tiktoken.get_encoding('o200k_base')


def show_split(word, enc):
    ids = enc.encode(word)
    pieces = [enc.decode([i]) for i in ids]
    print(f'{word!r:18s} -> {len(ids)} token(s): {pieces}')


# Words that actually carry a diacritic in the passage
diacritic_words = ['gósto', 'são', 'não', 'Estremeço', 'até', 'perfeição']

# Archaic-spelled words missing a diacritic modern Portuguese would use
archaic_words = ['tocaveis', 'visiveis', 'especie', 'verbaes', 'pagina',
                  'inattingivel', 'delirio', 'syntactica']

print('=== Diacritic-bearing words ===\n')
for w in diacritic_words:
    print('GPT-4:')
    show_split(w, gpt4)
    print('GPT-4o:')
    show_split(w, gpt4o)
    print()

print('=== Archaic-spelled (no diacritic) words ===\n')
for w in archaic_words:
    print('GPT-4:')
    show_split(w, gpt4)
    print('GPT-4o:')
    show_split(w, gpt4o)
    print()

# Test paginate for section 5 of week 1 assignemnt
print('\n=== Testing shared-root hypothesis for "pagina" ===\n')
print("Hypothesis: 'pagina' is a token because it's a substring of English")
print("'paginate'/'pagination'. Test: does it show up as a piece in those words?\n")
for w in ['paginate', 'pagination', 'paginated']:
    print('GPT-4:')
    show_split(w, gpt4)
```
